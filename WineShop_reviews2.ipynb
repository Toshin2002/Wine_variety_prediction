```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize

nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')

lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))

def preprocess_text(text):
    tokens = word_tokenize(text.lower())  # Tokenize the text
    filtered_tokens = [lemmatizer.lemmatize(token) for token in tokens if token.isalpha() and token not in stop_words]
    return ' '.join(filtered_tokens)

df_trail['review_description'] = df_trail['review_description'].apply(preprocess_text)
```

```python
print(df_trail.head(10))
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df_trail['variety_encoded'] = le.fit_transform(df_trail['variety'])
df_trail
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vect = TfidfVectorizer(analyzer='word',stop_words= 'english',max_features=5000)
new_df = vect.fit_transform(df_trail['review_description'])
new_df
```

```python
df_desc= new_df.toarray()
len(df_desc)
```
